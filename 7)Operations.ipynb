```python
#filtering

students = pd.DataFrame({
    "Name":["A","B","C","D"],
    "Marks":[70,85,90,60]
})
```

```python
students[students['Marks']>70]
```

```python
#multiple conditions
students[(students['Marks']>80) & (students['Marks']<90)]


```

```python
#assending sorting
students.sort_values('Marks',ascending=True)
```

```python
#descending sorting
students.sort_values('Marks', ascending=False)
```
