```python
import numpy as np
import pandas as pd
```

```python
data=pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)

})
data
```

```python
# Group by Category and calculate the sum of Sales
c=data.groupby('Category')['Sales'].sum()
c
```

```python
# Group by Store and calculate the sum of Sales
c=data.groupby('Store')['Sales'].sum()
c
```

```python
# Group by multiple columns
# Group by Category and Store
c=data.groupby(['Category','Store'])['Sales'].sum()
c
```

```python
#numeric_only
#numeric_only is used when a dataframe contains both numeric and non-numeric columns, but you want an operation like mean(),sum()etc . but in modern versions of pandas , this usually does not need bcz non-numeric columns are automatically ignored.
```

```python
data.groupby('Sales').count()
# difference b/w count() and size()
# count() = counts non null values column by column
# size()= counts total rows,including rows with missing values

```
